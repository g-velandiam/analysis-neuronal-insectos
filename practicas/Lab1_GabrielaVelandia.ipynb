## 2. Listas: varios valores juntos

Aquí, los instantes (en segundos) en que una neurona disparó un *spike* (potencial de acción):

````python
tiempos_spikes = [0.12, 0.35, 0.51, 0.98, 1.20, 1.45, 1.88]
print(tiempos_spikes)
````

## 3. Funciones ya incluidas en Python

Una **función** recibe algo (los *argumentos*, entre paréntesis), hace un trabajo y devuelve un resultado.
Patrón: `nombre_funcion(argumento)`. Estas ya vienen listas en Python:

````python
tiempos_spikes = [0.12, 0.35, 0.51, 0.98, 1.20, 1.45, 1.88]

print(len(tiempos_spikes))   # cuantos spikes hubo
print(min(tiempos_spikes))   # menor tiempo de spike
print(max(tiempos_spikes))   # mayor tiempo de spike
print(sum(tiempos_spikes))   # suma de todos los tiempos
````

## 4. Escribe tu propia función

Con `def` empaquetas una operación para reutilizarla. Definamos la **tasa de disparo** (spikes por segundo).
Fíjate en la **sangría** de las líneas de adentro: es obligatoria y Colab la pone sola al presionar Enter después de `:`.

````python
def tasa_de_disparo(tiempos, duracion):
    n = len(tiempos)          # cuenta los spikes
    return n / duracion       # los divide por la duracion

tiempos_spikes = [0.12, 0.35, 0.51, 0.98, 1.20, 1.45, 1.88]
tasa = tasa_de_disparo(tiempos_spikes, max(tiempos_spikes))
print("Tasa de disparo:", tasa, "spikes/segundo")
````

### Repetir acción con un ciclo `for`

Un ciclo `for` repite una acción para cada elemento de una lista. Lo veremos a fondo más adelante:

````python
tiempos_cortos = [0.12, 0.35, 0.51]

for tiempo in tiempos_cortos:
    print("Spike en", tiempo, "segundos")
````

# Bloque B · Diccionarios y tablas de datos

## 5. Diccionarios: datos con etiqueta

Una **lista** guarda valores en orden y los buscas por posición (`lista[0]`).
Un **diccionario** guarda valores con **nombre** y los buscas por esa etiqueta (`dicc["nombre"]`).

Se escribe entre llaves `{ }`, con parejas `llave: valor`.
Es la forma natural de describir **un** registro experimental: qué animal, qué neurona, qué estímulo.

````python
registro = {
    "especie": "Acheta domestica",
    "neurona": "N1",
    "estimulo": "viento",
    "duracion_s": 2.0,
    "tiempos_spikes": [0.12, 0.35, 0.51, 0.98, 1.20]
}

print(registro["especie"])          # se busca por la llave, no por la posicion
print(registro["tiempos_spikes"])   # un valor puede ser una lista completa
````

Puedes ver qué llaves tiene, agregar nuevas y haceer cálculos con sus valores:

````python
print(registro.keys())     # todas las etiquetas
print(registro.values())   # todos los valores

# agregar una llave nueva: se calcula la tasa y se guarda dentro del mismo registro
registro["tasa_hz"] = len(registro["tiempos_spikes"]) / registro["duracion_s"]

print(registro["tasa_hz"], "spikes/segundo")
````

### De un registro a muchos

Un experimento no es un registro: son decenas. Si pones **varios diccionarios dentro de una lista**,
ya tienes la estructura de una tabla — cada diccionario es una fila, cada clave es una columna.

````python
registros = [
    {"neurona": "N1", "estimulo": "viento",   "n_spikes": 12, "duracion_s": 2.0},
    {"neurona": "N1", "estimulo": "silencio", "n_spikes": 3,  "duracion_s": 2.0},
    {"neurona": "N2", "estimulo": "viento",   "n_spikes": 18, "duracion_s": 2.0},
    {"neurona": "N2", "estimulo": "silencio", "n_spikes": 5,  "duracion_s": 2.0},
]

for r in registros:                 # recorre fila por fila
    tasa = r["n_spikes"] / r["duracion_s"]
    print(r["neurona"], "-", r["estimulo"], "->", tasa, "Hz")
````

Funciona, pero fíjate en el trabajo que cuesta: un ciclo, una cuenta a mano, texto suelto.
Con 4 filas se puede; con 4000 no. Para eso existe `pandas`.

## 6. `pandas`: la tabla como objeto

Su objeto central se llama **DataFrame**: piénsalo como una hoja de cálculo que se maneja con código.

Primero **importamos** la librería. `as pd` es solo un apodo para escribir menos.

````python
import pandas as pd

df = pd.DataFrame(registros)   # la lista de diccionarios se convierte en tabla
df
````

Una línea reemplazó al ciclo `for`. Y crear columnas nuevas se hace de una vez para **todas** las filas:

````python
df["tasa_hz"] = df["n_spikes"] / df["duracion_s"]   # no hay ciclo: opera sobre la columna entera
df
````

### 6.1 Cargar datos desde un archivo

En el laboratorio los datos llegan en archivos, casi siempre en formato **CSV**
(*comma-separated values*: texto plano donde las columnas se separan por comas).

La celda de abajo **fabrica** un archivo CSV de ejemplo para que practiques.
No necesitas entenderla todavía; ejecútala y sigue. Simula 3 neuronas del sistema cercal de un grillo
registradas con y sin estímulo de viento.

````python
# --- Solo para generar datos de practica. Ejecuta y sigue. ---
import random
random.seed(7)

filas = []
for neurona, base in [("N1", 4), ("N2", 7), ("N3", 2)]:
    for estimulo, ganancia in [("silencio", 1.0), ("viento", 3.5)]:
        for ensayo in range(1, 21):
            n = max(0, int(random.gauss(base * ganancia, base * 0.6)))
            filas.append({
                "neurona": neurona,
                "estimulo": estimulo,
                "ensayo": ensayo,
                "n_spikes": n,
                "duracion_s": 2.0,
                "amplitud_mV": round(random.gauss(0.45, 0.08), 3)
            })

pd.DataFrame(filas).to_csv("registro_grillo.csv", index=False)
print("Archivo creado: registro_grillo.csv")
````

Ahora lo leemos. `pd.read_csv()` es probablemente la función que más vas a usar en este curso.

````python
datos = pd.read_csv("registro_grillo.csv")
datos.head()      # muestra solo las primeras 5 filas
````

### 6.4 Filtrar: quedarse con las filas que cumplen una condición

Esta es la operación más importante del bloque. Se hace en dos pasos:

1. Escribes una **condición** sobre una columna → obtienes una columna de `True`/`False` (booleana).
2. Metes esa condición dentro de `datos[...]` → te quedas solo con las filas `True`.

````python
condicion = datos["estimulo"] == "viento"    # paso 1: verdadero/falso por fila
print(condicion.head())

viento = datos[condicion]                    # paso 2: filtra
print()
print("Filas con viento:", viento.shape[0])
viento.head()
````

Normalmente se escribe todo en una línea. Para combinar condiciones se usa `&` (**y**) ; `|` (**o**) ; `>` (**mayor**) ; `<` (**menor**).
**Cada condición va entre paréntesis**:

````python
# una sola neurona, solo con viento
n2_viento = datos[(datos["neurona"] == "N2") & (datos["estimulo"] == "viento")]
print(n2_viento.shape)
n2_viento.head()
````

````python
# ensayos con respuesta alta
altos = datos[datos["n_spikes"] > 25]
print("Ensayos con mas de 25 spikes:", len(altos))
altos.head()
````

### 6.5 Agrupar: resumir por condición

`groupby` responde la pregunta central de casi cualquier experimento:
*¿el promedio cambia entre grupos?*

Se lee como una frase: **agrupa por** `estimulo`, **toma** `n_spikes`, **calcula** el promedio.

````python
datos.groupby("estimulo")["n_spikes"].mean()
````

````python
# se puede agrupar por dos columnas a la vez
resumen = datos.groupby(["neurona", "estimulo"])["n_spikes"].agg(["mean", "std", "count"])
resumen
````

````python
datos["estimulo"].value_counts()    # cuantas filas hay de cada categoria
````

### 6.6 Cerrar el ciclo: de la tabla a la figura

Todo lo anterior existe para llegar aquí.

matplotlib es la librería de gráficos de Python: la que realmente dibuja píxeles en la pantalla. pandas no sabe dibujar — cuando le pides un gráfico, se lo encarga a matplotlib por debajo.

Por eso la celda de abajo usa las dos:

````python
import matplotlib.pyplot as plt   # matplotlib es quien dibuja; plt es el apodo estandar

promedios = datos.groupby("estimulo")["n_spikes"].mean()   # pandas: prepara los datos

promedios.plot(kind="bar", color=["#9db8c9", "#c99d9d"])   # pandas pide el grafico, matplotlib lo dibuja
plt.ylabel("Spikes por ensayo (promedio)")   # de aqui en adelante, todo es matplotlib
plt.xlabel("Estimulo")
plt.title("Respuesta cercal al viento")
plt.xticks(rotation=0)                       # texto del eje x horizontal
plt.show()                                   # muestra la figura terminada
````

### Reto integrador — Tu primer mini-pipeline

En **una sola celda**, y usando prompts si lo necesitas, escribe un análisis completo:

1. Lee `registro_grillo.csv`.
2. Quédate solo con las filas de estímulo `viento`.
3. Calcula la tasa de disparo (`n_spikes / duracion_s`) en una columna nueva `tasa_hz`.
4. Obtén el promedio de `tasa_hz` por neurona.
5. Grafícalo como barras, con labels y título.

Esto es, en miniatura, la estructura de tu proyecto final: **dato crudo → filtro → medida → figura**.

````python
# TODO: tu mini-pipeline completo aqui

import pandas as pd
import matplotlib.pyplot as plt

registro_g = pd.read_csv("registro_grillo.csv")

condicion4 = registro_g["estimulo"] == "viento"

viento_si = registro_g[condicion4]

viento_si["tasa_hz"] = viento_si["n_spikes"] / viento_si["duracion_s"]

prom_g = viento_si.groupby("neurona")["tasa_hz"].mean()

prom_g.plot(kind="bar", color="lightblue", edgecolor="gray")

plt.xlabel("Neurona")

plt.ylabel("Tasa de disparo promedio (Hz)")

plt.title("Tasa de disparo por neurona")

plt.show()

````
